```python
from pathlib import Path
import numpy as np
from tqdm import tqdm
```

```python
def count_atoms_in_xyz(file_path):
    """
    Conta il numero di atomi in un file XYZ.
    
    :param file_path: Percorso del file XYZ.
    :return: Numero di atomi.
    """
    try:
        with open(file_path, 'r') as file:
            # La prima riga contiene il numero di atomi
            num_atoms = int(file.readline().strip())
            return num_atoms
    except Exception as e:
        print(f"Errore durante la lettura del file: {e}")
        return None
```

```python
path = Path("/home/tommaso/xyz_files_G")

min = np.inf
max = -np.inf

samples =[ f for f in path.iterdir() if f.suffix.lower() ==".xyz"]
for sample in tqdm(samples):
    atoms = count_atoms_in_xyz(str(sample))
    if atoms > max:
        max = atoms
    if atoms < min:
        min = atoms
```
