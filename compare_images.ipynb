```python
from PIL import Image
import numpy as np

def compare_images(image_path1, image_path2):
    # Apri le immagini
    image1 = Image.open(image_path1)
    image2 = Image.open(image_path2)
    
    # Converti le immagini in array di NumPy
    np_image1 = np.array(image1)
    np_image2 = np.array(image2)
    
    # Controlla se le dimensioni delle immagini sono diverse
    if np_image1.shape != np_image2.shape:
        return True  # Le immagini sono diverse
    
    # Confronta gli array pixel per pixel
    difference = np_image1 - np_image2
    if np.any(difference):
        return True  # Le immagini sono diverse
    
    return False  # Le immagini sono uguali

# Esempio di utilizzo
image_path1 = '/home/tommaso/git_workspace/GrapheNet/not_published/UMAP/25_clusters/cluster_2/test/neutral_495.png'
image_path2 = '/home/tommaso/git_workspace/GrapheNet/data_final/UMAP/cluster_2/training_dataset/test/neutral_495.png'

if compare_images(image_path1, image_path2):
    print("Le immagini sono diverse.")
else:
    print("Le immagini sono uguali.")
```
